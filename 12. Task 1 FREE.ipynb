### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
x1 = int(input("Enter your first coordinate x1 : "))
x2 = int(input("Enter your second coordinate x2 : "))
y1 = int(input("Enter your first coordinate y1 : "))
y2 = int(input("Enter your second coordinate y2 : "))

euc = ((x2-x1)**2 + (y2-y1)**2)**0.5
print(f"The Euclidean distance between these two points are : {euc}")
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
principle = float(input("Enter your prnciple amount : "))
rate = float(input("Enter your rate of interest : " ))
time = float(input("Enter the time period in years : "))

simple_interest = (principle * rate * time)/100

print(f"The simple interest is : {simple_interest}")
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
legs = int(input("Enter the number of the legs :  "))
heads = int(input("Enter the number of heads : "))

dogs = (legs - 2* heads)/2
chicken = heads - dogs

print(f"The number of dogs are : {dogs} and the number of chickens are : {chicken}")
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
n  = int(input("Enter the number of terms : "))
sum = 0
for i in range(1,n+1):
    i = i**2
    sum += i
print(f"The sum of the series is : {sum}")
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
n = int(input("Enter a number : "))
a1 = int(input("Enter a number : "))
a2 = int(input("Enter a number : "))

d = a2 - a1

arth = a1 + (n-1) *  d

print(f"The arithmetic series for the {n}th term is {arth} ")
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here

l = int(input("Enter the length of tank : "))
b = int(input("Enter the breadth of tank : "))
h = int(input("Enter the height of tank : "))
hc = int(input("Enter the height of circle : "))
r = int(input("Enter the radius of circle : "))

volume_of_tank = h * l * b
print(f"The volume of tank is  {volume_of_tank} cm³ ")

volume_of_glass = 3.14 * r**2 * hc
print(f"The volume of glass is {volume_of_glass} cm³")

glasses_of_milk_obtained = volume_of_tank//volume_of_glass

print(f"Total complete glasses of milk can be obtained from {volume_of_tank} cm³ tank using {volume_of_glass} cm³ glass are {glasses_of_milk_obtained} ")
````
