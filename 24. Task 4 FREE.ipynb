###`Problem 1:` Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

````python
# Write code here

list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

emp_list = []

for i in range (len(list1)):
    emp_list.append([list1[i], list2[i]])
print(emp_list)



````

###`Problem 7:` Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

`Input:`
```
['1ac21', '23fg', '456', '098d','1','kls']
```4

`Output:`
```
['456', '23fg', '1ac21', '1', 'kls', '098d']

```

````python
lst = ['1ac21', '23fg', '456', '098d', '1', 'kls']

emp_list = []

for i in lst:
    product = 1

    for j in i:
        if j.isdigit():
            product = product * int(j)

    emp_list.append([product, i])

emp_list.sort(reverse=True)

result = []

for i in emp_list:
    result.append(i[1])

print(result)
````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

````python
h = ['campusxIs', 'bestFor', 'dataScientist']
emp_list = []

for i in h:
    word = ""

    for j in i:
        if j.isupper():
            emp_list.append(word)
            word = j
        else:
            word += j

    emp_list.append(word)

print(emp_list)
````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:
```bash
[1,2,3,4,5,7,8]
```

````python
# Write code here

lst1 = [1,2,3,4,5,1]
lst2 = [2,3,5,7,8]

lst3 = set(lst1)
lst4 = set(lst2)

lst5 = lst3.union(lst4)
print(list(lst5))
````

### `Problem 12:` Write a program that can find the max number of each row of a matrix

**Example:**

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

````python
# Write code here
lst = [[1,2,3],[4,5,6],[7,8,9]]
empty = []

for i in lst:
    for j in i:
        j = max(i)
    empty.append(j)
print(empty)


````

### `Problem 14:` Write a list comprehension that can transpose a given matrix

matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

[1, 4, 7]<br>
[2, 5, 8]<br>
[3, 6, 9]<br>

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transpose = [[row[i] for row in matrix] for i in range(len(matrix[0]))]

print(transpose)
````
