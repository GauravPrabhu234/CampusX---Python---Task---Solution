### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
salary = int(input("Enter your salary: "))
hra = salary * 10/100
print(hra)
da = salary * 5/100
print(da)
pf = salary * 3/100
print(pf)

if salary < 500000:
  tax = 0
elif 500000 <= salary <= 1000000:
  tax = 0.10
elif 1000000 <= salary <= 2000000:
  tax = 0.20
elif salary > 2000000:
  tax = 0.30

annual_salary = salary - hra - da - pf
annual_salary = annual_salary - (annual_salary * tax)
monthly_salary = annual_salary / 12

print(f"Your monthly salary is {monthly_salary}")
````

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:


### `Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
x = 0
y = 0

while True:
    movement = input("Enter movement: ")

    if movement == "!":
        break

    direction, steps = movement.split()
    steps = int(steps)

    if direction == "UP":
        y += steps
    elif direction == "DOWN":
        y -= steps
    elif direction == "LEFT":
        x -= steps
    elif direction == "RIGHT":
        x += steps

distance = round((x**2 + y**2) ** 0.5)
print(distance)
````

### `Problem 12`:Write a program to print whether a given number is a prime number or not

````python
n = int(input("Enter a number: "))

if n <= 1:
    print(f"{n} is not a prime number.")
else:
    for i in range(2, n):
        if n % i == 0:
            print(f"{n} is not a prime number.")
            break
    else:
        print(f"{n} is a prime number.")
````

### `Problem 13`:Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
n = int(input("Enter a number: "))
original = n
total = 0
digits = len(str(n))

while n > 0:
    digit = n % 10
    total = total + digit ** digits
    n = n // 10

if total == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
````

### `Problem 14`:Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
# Write code here
Hour = int(input("Enter the hour : "))
Minute = int(input("Enter the minute : "))

hour_diff = (Hour%12) * 30 
minute_diff = Minute * 0.5
a = hour_diff + minute_diff

min_hand = Minute * 6

difference = abs(a - min_hand)

min_angle = min(difference,360 - difference)

print(int(min_angle))
````
