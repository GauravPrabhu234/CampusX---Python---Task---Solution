### **`Problem-1:`** Write a Python function that takes a list and returns a new list with unique elements of the first list.

**Exercise 1:**

Input:

```bash
[1,2,3,3,3,3,4,5]
```

Output:

```bash
[1, 2, 3, 4, 5]
```

````python
def lis1(list1):
    new_list = []
    for i in list1:
        if i not in new_list:
            new_list.append(i)
    return new_list

lis1([1,2,4,5,6,7,7,8,9,9,9,8,7])

    
````

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
# Write code here

def sort_words(n):
    lst = n.split("-")
    lst.sort()
    return "-".join(lst)

l = sort_words("green-red-yellow-black-white")
print(l)
    
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Write code here

def count_up(n):
    up_count = 0
    low_count = 0
    for i in n:
        if i.isupper():
            up_count +=1
        elif i.islower():
            low_count +=1
    return up_count , low_count

count_up('CampusX is an Online Mentorship Program fOr EnginEering studentS.')
````

### **`Problem 4:`** Write a Python program to print the even numbers from a given list.
```
Sample List : [1, 2, 3, 4, 5, 6, 7, 8, 9]
Expected Result : [2, 4, 6, 8]
```

````python
# Write code here

def is_even(n):
    new_lst = []
    for i in n:
        if i%2 == 0:
            new_lst.append(i)
    return new_lst

is_even([1,2,3,4,5,6,7,8])
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here

def perfect(n):
    total = 0
    for i in range(1,n):
        if  n%i == 0:
            total += i
    if total == n:
        return print("It is a perfect number.")
    else:
        return print("Not a perfect number ")

perfect(28)     
    
            
````

`Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
# Write code here

# str1 = 'hello how are you i am fine thank you'




def max_word_count(str1):
    new_str = str1.split()
    new_dict = {}

    for word in new_str:
        if word in new_dict:
            new_dict[word] +=1
        else:
            new_dict[word] = 1

    max_count = 0
    max_word = ""

    for key,value in new_dict.items():
        if value > max_count:
            max_count = value
            max_word = key
    return max_word , max_count



max_word_count('hello how are you i am fine thank you')



````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
# Write code here

def hist(lst):
    bins = {}
    for i in lst:
        if 0 <= i <= 10:
            bin_name = "0-10"
            if bin_name in bins:
                bins[bin_name] += 1
            else:
                bins[bin_name] = 1           
        elif 11 <= i <= 20:
            bin_name = "11 - 20"
            if bin_name in bins:
                bins[bin_name] += 1
            else:
                bins[bin_name] = 1
        elif 21 <= i <= 30:
            bin_name = "21 - 30"
            if bin_name in bins:
                bins[bin_name] += 1
            else:
                bins[bin_name] = 1
        elif 31 <= i <= 40:
            bin_name = "31 - 40"
            if bin_name in bins:
                bins[bin_name] += 1
            else:
                bins[bin_name] = 1
        elif 41 <= i <= 50:
            bin_name = "41 - 50"
            if bin_name in bins:
                bins[bin_name] += 1
            else:
                bins[bin_name] = 1
    return bins

hist([9,13,42,15,37,22,39,41,50])
````

`Problem-9` Write a python function that accepts a list of 2D co-ordinates and a query point, and then finds the the co-ordinate which is closest in terms of distance from the query point.

```
List of Coordinates
[(1,1),(2,2),(3,3),(4,4)]
Query Point
(0,0)
```

```
Output
Nearest to (0,0) is (1,1)
```

````python
# Write code here
import math

def nearest_coordinate(coordinates, query):
    min_distance = float('inf')
    nearest_point = None

    for point in coordinates:
        x1, y1 = point
        x2, y2 = query

        distance = math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

        if distance < min_distance:
            min_distance = distance
            nearest_point = point

    return nearest_point


coordinates = [(1, 1), (2, 2), (3, 3), (4, 4)]
query = (0, 0)

result = nearest_coordinate(coordinates, query)
print(f"Nearest to {query} is {result}")
````

`Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
def bow(sentences):
    vocabulary = []

    # Step 1: Create vocabulary
    for sentence in sentences:
        words = sentence.lower().split()

        for word in words:
            if word not in vocabulary:
                vocabulary.append(word)

    # Step 2: Create Bag of Words
    bow_dict = {}

    for sentence in sentences:
        words = sentence.lower().split()
        frequency = {}

        for word in vocabulary:
            frequency[word] = words.count(word)

        bow_dict[sentence] = frequency

    return vocabulary, bow_dict


sentences = [
    "I love Python",
    "I love Machine Learning",
    "Python is great"
]

vocabulary, result = bow(sentences)

print("Vocabulary:", vocabulary)
print("Bag of Words:")

for sentence, frequency in result.items():
    print(sentence, ":", frequency)
````
